## 4. INE: gráficas y confección por municipio (prueba)

El INE no publica las subclases CAE (5 cifras) por municipio. Aquí se comprueba si publica las divisiones (2 cifras) por municipio: la 18 (impresión) y la 14 (industria del vestuario). Primero se leen los metadatos para encontrar los códigos.

**Cómo funciona**

- **Qué hace:** lee los metadatos del indicador de empresas por división y busca los códigos de las divisiones que usamos.
- **Por qué así:** el INE numera las dimensiones (`Dim1`, `Dim3`…); hay que encontrar cuál es la actividad económica. `next(...)` devuelve el primer elemento que cumple la condición.
- **Herramientas:** list comprehension anidada, `startswith`, `pd.DataFrame`.
- **Nuevo (no visto en clase):** `next` con un generador.

```python
ine = config["ine"]
indicador_div = ine["indicador_empresas_division"]
ruta_meta_div = RUTA_RAW / f"ine_metadatos_{indicador_div}.json"
if not ruta_meta_div.exists():
    recoleccion.guardar_json(recoleccion.pedir_metadatos_ine(indicador_div, ine), ruta_meta_div)
meta_div = recoleccion.leer_json(ruta_meta_div)
meta_div = meta_div[0] if isinstance(meta_div, list) else meta_div
print(meta_div["IndicadorNome"])
print("Último año:", meta_div["UltimoPeriodo"])

# Qué dimensión es la actividad económica y qué códigos tienen las divisiones 18 y 14
dimensiones = meta_div["Dimensoes"]["Descricao_Dim"]
dim_actividad = next(d["dim_num"] for d in dimensiones if d["abrv"].startswith("Atividade"))
categorias = [c for grupo in meta_div["Dimensoes"]["Categoria_Dim"] for lista in grupo.values() for c in lista]
# Las divisiones que usamos están en config.yaml (ine > divisiones_cae)
divisiones = [c for c in categorias if c["dim_num"] == dim_actividad and c["categ_cod"] in ine["divisiones_cae"]]
pd.DataFrame(divisiones)[["categ_cod", "categ_dsg"]]
```

**Cómo funciona**

- **Qué hace:** pide los datos de cada división y comprueba si hay dato por municipio (código de 7 caracteres).
- **Por qué así:** era una prueba: si el INE publica las divisiones por municipio, la competencia se puede medir por municipio y no solo por región. Resultado: sí.
- **Herramientas:** set comprehension, `len`, f-strings.

```python
periodo_div = "S7A" + str(meta_div["UltimoPeriodo"])
respuestas_div = recoleccion.recoger_ine_por_categorias(
    indicador_div, [c["categ_cod"] for c in divisiones], ine, periodo_div, RUTA_RAW,
    dimension=f"Dim{dim_actividad}",
)

# ¿Hay datos por municipio? En el INE, el código de un municipio tiene 7 caracteres (por ejemplo 16E0603)
for codigo, respuesta in respuestas_div.items():
    primero = respuesta[0] if isinstance(respuesta, list) else respuesta
    registros = [r for lista in primero.get("Dados", {}).values() for r in lista]
    con_dato = {r["geodsg"] for r in registros if len(str(r["geocod"])) == 7}
    print(f"División {codigo}: {len(registros)} registros; municipios con dato: {len(con_dato)}")
```

## 5. Portal BASE: anuncios de procedimientos

Los concursos públicos se anuncian antes de adjudicarse y cualquier empresa puede presentarse. Este conjunto de datos (mismo portal, se actualiza cada dos semanas) permite ver cuántos hubo de ropa en la región y, más adelante, vigilar los que estén abiertos.

**Cómo funciona**

- **Qué hace:** lista los archivos del dataset de anuncios del Portal BASE con su tamaño.
- **Por qué así:** igual que con los contratos: antes de descargar se mira qué hay.
- **Herramientas:** `pd.to_numeric`, `sort_values`.

```python
ruta_info_anuncios = RUTA_RAW / "base_anuncios_info.json"
if ruta_info_anuncios.exists():
    info_anuncios = recoleccion.leer_json(ruta_info_anuncios)
else:
    info_anuncios = recoleccion.listar_recursos_dataset(config["base"]["api_anuncios"])
    recoleccion.guardar_json(info_anuncios, ruta_info_anuncios)

recursos_anuncios = pd.DataFrame(info_anuncios["resources"])
recursos_anuncios["mb"] = (pd.to_numeric(recursos_anuncios["filesize"], errors="coerce") / 1_000_000).round(1)
recursos_anuncios[["title", "format", "mb"]].sort_values("title")
```

**Cómo funciona**

- **Qué hace:** descarga el archivo de anuncios de cada año, en Excel si existe y, si no, en CSV.
- **Por qué así:** `sort_values(key=...)` pone primero el Excel; `continue` salta los años sin archivo; `if ruta.exists()` evita descargar dos veces.
- **Herramientas:** filtros con `&`, `isin`, `iloc`, `continue`.
- **Nuevo (no visto en clase):** `sort_values` con `key`.

```python
# Descarga de los anuncios de los años elegidos (en Excel si existe; si no, en CSV)
for ano in config["base"]["anos"]:
    candidatos = recursos_anuncios[
        recursos_anuncios["title"].str.contains(str(ano))
        & recursos_anuncios["format"].str.lower().isin(["xlsx", "csv"])
    ].sort_values("format", key=lambda formato: formato.str.lower() != "xlsx")
    if candidatos.empty:
        print(ano, ": no hay ningún archivo con ese año en el título")
        continue
    fila = candidatos.iloc[0]
    nombre = fila["title"] if "." in fila["title"] else f"{fila['title']}.{fila['format'].lower()}"
    ruta = RUTA_RAW / "base_anuncios" / nombre
    if ruta.exists():
        print(nombre, "ya estaba descargado")
    else:
        print("Descargando", nombre, fila["mb"], "MB")
        recoleccion.descargar_archivo(fila["url"], ruta)
```

## 6. Clubes: ¿publican las asociaciones distritales de fútbol la lista de sus clubes?

Primero se comprueba el robots.txt y se guardan la portada y las páginas que hablan de clubes para revisarlas.

**Resultado:** las webs de Viseu, Guarda y Castelo Branco devolvieron error, y en la de Coimbra las páginas de clubes dieron error o el robots.txt no las permite. No se insiste: el plan B son los futbolistas federados por distrito del INE (sección 9).

**Cómo funciona**

- **Qué hace:** comprueba el robots.txt de las webs de las asociaciones distritales de fútbol, descarga la portada y busca enlaces que hablen de clubes.
- **Por qué así:** el robots.txt dice qué páginas permite descargar un sitio: si no lo permite, no se descarga. `try/except` sigue con el siguiente distrito si una web falla.
- **Herramientas:** `try/except`, `continue`, `BeautifulSoup` (dentro de `enlaces_con_palabras`).
- **Nuevo (no visto en clase):** `urllib.robotparser` (lector de robots.txt).

```python
import time

web = config["web"]
carpeta_clubes = RUTA_RAW / "clubes"
enlaces_af = {}
for distrito, url in web["asociaciones_futbol"].items():
    ruta = carpeta_clubes / f"af_{recoleccion.nombre_en_url(distrito)}.html"
    try:
        if not ruta.exists() and not recoleccion.permitido_por_robots(url, web["user_agent"]):
            print(distrito, ": el robots.txt no permite descargar la página")
            continue
        html = recoleccion.descargar_html(url, ruta, web["user_agent"])
    except Exception as error:
        print(distrito, ": no se pudo descargar", type(error).__name__)
        continue
    enlaces_af[distrito] = recoleccion.enlaces_con_palabras(html, url, ["clube", "filiad", "associad"])
    print(f"{distrito}: {len(enlaces_af[distrito])} enlaces sobre clubes")
recoleccion.guardar_json(enlaces_af, carpeta_clubes / "enlaces_af.json")
```

**Cómo funciona**

- **Qué hace:** guarda las 5 primeras páginas sobre clubes de cada asociación, respetando el robots.txt y con una pausa entre peticiones.
- **Por qué así:** se guardan para revisarlas a mano antes de decidir si se pueden extraer los clubes.
- **Herramientas:** `enumerate(start=1)`, slicing, `try/except`, `time.sleep`.

```python
# Se guardan las primeras 5 páginas sobre clubes de cada asociación, para revisarlas
for distrito, enlaces in enlaces_af.items():
    for numero, enlace in enumerate(enlaces[:5], start=1):
        es_pdf = enlace["url"].lower().endswith(".pdf")
        ruta = carpeta_clubes / f"af_{recoleccion.nombre_en_url(distrito)}_{numero}.{'pdf' if es_pdf else 'html'}"
        if ruta.exists():
            continue
        try:
            if not recoleccion.permitido_por_robots(enlace["url"], web["user_agent"]):
                print(distrito, numero, ": no permitido por el robots.txt")
                continue
            if es_pdf:
                recoleccion.descargar_archivo(enlace["url"], ruta)
            else:
                recoleccion.descargar_html(enlace["url"], ruta, web["user_agent"])
            time.sleep(web["pausa_segundos"])
        except Exception as error:
            print(distrito, numero, ": no se pudo descargar", type(error).__name__)
print("Archivos guardados:", len(list(carpeta_clubes.glob("af_*"))))
```

## 7. Contactos de las cámaras (CCDR Centro)

La CCDR Centro tiene una página de contacto por municipio (solo de la región Centro).

**Resultado:** el robots.txt no permite descargarlas, así que no se descarga nada. El contacto de las cámaras se busca en su web oficial.

**Cómo funciona**

- **Qué hace:** comprueba si el robots.txt de la CCDR Centro permite descargar las páginas de contacto de las cámaras y, solo si lo permite, las descarga.
- **Por qué así:** no lo permite, así que no se descarga nada. Es una decisión ética: no se buscan rodeos.
- **Herramientas:** `if`, `format` en una URL, `try/except`.

```python
import time

web = config["web"]
permitido = recoleccion.permitido_por_robots(web["url_contactos_ccdrc"].format(nombre="tabua"), web["user_agent"])
print("¿Lo permite el robots.txt?", permitido)

if permitido:
    sin_pagina = []
    for municipio in municipios:
        nombre = recoleccion.nombre_en_url(municipio)
        ruta = RUTA_RAW / "contactos_ccdrc" / f"{nombre}.html"
        if ruta.exists():
            continue
        try:
            recoleccion.descargar_html(web["url_contactos_ccdrc"].format(nombre=nombre), ruta, web["user_agent"])
            time.sleep(web["pausa_segundos"])
        except Exception:
            sin_pagina.append(municipio)
    print("Municipios sin página en la CCDR Centro:", sin_pagina)
```

## 8. Presupuesto del Estado 2026: lo que recibe cada freguesia

La Lei 73-A/2025 (OE2026) debía incluir el Fondo de Financiación de las Freguesias de cada una, para medir el tamaño de cada junta.

**Resultado:** el PDF se descargó, pero ese mapa no viene en él; el que sí se pudo leer (transferencia de competencias, DL 57/2019) solo cubre 37 de los 66 municipios. Se descartó y se explica en el README.

**Cómo funciona**

- **Qué hace:** descarga el PDF de la ley del Presupuesto del Estado 2026, si el robots.txt lo permite.
- **Por qué así:** se buscaba el dinero que recibe cada freguesia para medir el tamaño de cada junta.
- **Herramientas:** `Path.stat().st_size`, `round`.

```python
web = config["web"]
ruta_oe = RUTA_RAW / "oe2026" / "lei-73-a-2025.pdf"
if not ruta_oe.exists():
    if recoleccion.permitido_por_robots(web["url_oe2026"], web["user_agent"]):
        recoleccion.descargar_archivo(web["url_oe2026"], ruta_oe)
    else:
        print("El robots.txt no permite descargar el PDF")
if ruta_oe.exists():
    print("OE2026 descargado:", round(ruta_oe.stat().st_size / 1_000_000, 1), "MB")
```
